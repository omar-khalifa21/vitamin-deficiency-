```python
df['alcohol_consumption'] = df['alcohol_consumption'].fillna("does not drink")
```

```python
df['gender'] = df['gender'].map({'Male': 0, 'Female': 1})

df['smoking_status'] = df['smoking_status'].map({'Current': 0, 'Former': 1, 'Never': 2})
df['alcohol_consumption'] = df['alcohol_consumption'].map({'does not drink': 0, 'Moderate': 1, 'Heavy': 2})

df['exercise_level'] = df['exercise_level'].map({'Sedentary': 0, 'Light': 1, 'Moderate': 2, 'Active': 3})
df['diet_type'] = df['diet_type'].map({'Vegan': 0, 'Pescatarian': 1, 'Vegetarian': 2, 'Omnivore': 3})

df['sun_exposure'] = df['sun_exposure'].map({'Low': 0, 'Moderate': 1, 'High': 2})
df['income_level'] = df['income_level'].map({'Low': 0, 'Middle': 1, 'High': 2})
df['latitude_region'] = df['latitude_region'].map({'Low': 0, 'Mid': 1, 'High': 2})
```

```python
df.head()
```

```python
df.info()
```

```python
all_symptoms = set()

for i, count in df['symptoms_list'].value_counts().items():
    symptoms = i.split(';')
    all_symptoms.update(symptoms)
    print(f"{i}: {count}")
```

```python
all_symptoms
```

```python
df['symptoms_list'] = df['symptoms_list'].fillna("None")
```

```python
#make columns for each symptom
for symptom in all_symptoms:
    df[symptom] = df['symptoms_list'].apply(lambda x: 1 if symptom in x.split(';') else 0)
```

```python
df.drop('symptoms_list', axis=1, inplace=True)
```
